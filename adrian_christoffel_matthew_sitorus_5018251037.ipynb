```python
# Tugas: Sensor Log Analyzer
# Nama  : Adrian
# Kelas : Pemrograman Dasar

# 1. List data sensor (minimal 20 nilai, ada yang negatif dan di atas ambang batas)
data_sensor = [
    25.5, 30.0, -5.0, 85.0, 22.1,
    90.5, 18.0, -10.2, 45.0, 78.0,
    82.3, 15.0, 29.4, -1.0, 88.0,
    34.0, 92.1, 10.0, 50.0, 27.5
]

# Batas untuk menentukan status ALARM
batas_alarm = 80.0

# Variabel untuk menyimpan hasil perhitungan
total_nilai = 0
jumlah_data_valid = 0
jumlah_alarm = 0

# Variabel min dan max diisi None dulu sebagai penanda awal
nilai_min = None
nilai_max = None

print("----------------------------------------")
print("          HASIL LOG SENSOR              ")
print("----------------------------------------")

# Menggunakan loop manual untuk mengecek satu per satu
for i in range(len(data_sensor)):
    nilai = data_sensor[i]
    nomor_sampel = i + 1

    # Abaikan jika nilai negatif (data tidak valid)
    if nilai < 0:
        print("Sampel", nomor_sampel, ":", nilai, "-> INVALID (Diabaikan)")
    else:
        # Menentukan status NORMAL atau ALARM
        if nilai > batas_alarm:
            status = "ALARM"
            jumlah_alarm = jumlah_alarm + 1
        else:
            status = "NORMAL"

        print("Sampel", nomor_sampel, ":", nilai, "-> Status:", status)

        # Tambahkan ke total untuk hitung rata-rata nanti
        total_nilai = total_nilai + nilai
        jumlah_data_valid = jumlah_data_valid + 1

        # Cari nilai minimum secara manual
        if nilai_min is None:
            nilai_min = nilai
        elif nilai < nilai_min:
            nilai_min = nilai

        # Cari nilai maksimum secara manual
        if nilai_max is None:
            nilai_max = nilai
        elif nilai > nilai_max:
            nilai_max = nilai

# Menghitung rata-rata
if jumlah_data_valid > 0:
    rata_rata = total_nilai / jumlah_data_valid
else:
    rata_rata = 0

print("----------------------------------------")
print("              RINGKASAN                 ")
print("----------------------------------------")
print("Jumlah data valid :", jumlah_data_valid)
print("Jumlah ALARM      :", jumlah_alarm)
print("Rata-rata nilai   :", round(rata_rata, 2))
print("Nilai Minimum     :", nilai_min)
print("Nilai Maksimum    :", nilai_max)
print("----------------------------------------")
```
