```python
import numpy as np
from scipy import optimize

def f(x):
    return x**2 - 16

root = optimize.brentq(f, 0, 6)
print(f"Akar dari persamaan pada interval [0, 6] adalah x = {root}")
```

```python
import numpy as np

A = np.array([[3, 1], [1, 2]])
B = np.array([7, 5])

sol = np.linalg.solve(A, B)

print(f"Solusi: x = {sol[0]}, y = {sol[1]}")
```

```python
residual_nonlinear = f(root)
print(f"Residual nonlinear (f(4.0)): {residual_nonlinear}")

residual_linear = np.dot(A, sol) - B
print(f"Residual sistem linear (A * sol - B): {residual_linear}")
```

```python
import numpy as np


A_singular = np.array([[3, 1], [6, 2]])
B_singular = np.array([7, 5])


sol_singular = np.linalg.solve(A_singular, B_singular)
print(sol_singular)
```
